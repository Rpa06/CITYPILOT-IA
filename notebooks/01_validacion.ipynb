## 1. API de datos públicos (obligatorio)
Con `requests`, recupera **Valencia → coordenadas → clima actual**. No uses lat/lon de memoria como respuesta final: debes sacarlas del JSON de Geocoding. Muestra ciudad, lat, lon, temperatura y precipitación. Comprueba qué haces si `results` está vacío.

En `docs/APIS_Y_CONTRATOS.md` tienes URLs y parámetros. Puedes poner los campos `name`, `count`, `countryCode`, `current`, `timezone` en diccionarios de parámetros.

```python
geo_url = "https://geocoding-api.open-meteo.com/v1/search"
meteo_url = "https://api.open-meteo.com/v1/forecast"

ciudad = "valencia"
# petición GET geocoding, extraer latitud/longitud
response = requests.get(geo_url+f"?name={ciudad}&count=1&language=es&format=json", timeout=5)
# Buscamos results
if response.status_code == 200:
    data = response.json()
    if "results" in data and len(data["results"]) > 0:
        result = data["results"][0]
        geo_loc = {
            "ciudad": result.get("name"),
            "pais":result.get("country"),
            "lat": result.get("latitude"),
            "lon": result.get("longitude")
        }

# petición GET meteorología con esas coordenadas

response = requests.get(meteo_url+f"?latitude={geo_loc["lat"]}&longitude={geo_loc["lon"]}&current_weather=true&timezone=auto", timeout=5)
if response.status_code == 200:
    data = response.json()
    if "current_weather" in data:
        current_weather = data["current_weather"]
        met = {
            "temperatura_c": current_weather.get("temperature"),
            "precipitacion_mm": current_weather.get("precipitation"),
            "codigo_tiempo": current_weather.get("weathercode"),
            "instante": current_weather.get("time")
        }


# TODO(ALUMNADO): imprimir resultado y dejar evidencia
print(geo_loc)
print("--------")
print(met)
```
